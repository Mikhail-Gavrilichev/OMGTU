```python
def generate_partitions(n):
    # массив для хранения разбиения
    p = [0] * n

    # первое разбиение: само число
    k = 0
    p[k] = n

    while True:
        # вывод текущего разбиения
        print(f"{n} = {'+'.join(map(str, p[:k + 1]))}")

        # ищем правый элемент > 1
        rem_val = 0

        while k >= 0 and p[k] == 1:
            rem_val += p[k]
            k -= 1

        # если все единицы — работа завершена
        if k < 0:
            return

        # уменьшаем найденный элемент
        p[k] -= 1
        rem_val += 1

        # заполняем справа максимально возможными значениями
        while rem_val > p[k]:
            p[k + 1] = p[k]
            rem_val -= p[k]
            k += 1

        # остаток
        p[k + 1] = rem_val
        k += 1


generate_partitions(10)
```
