```python
from collections import deque

A = [1, 8, 4, 3, 5, 7, 4, 5, 6, 7]
k = 3

tail = deque()  # для вставки справа
head = deque()  # для обратной стороны

result = []

for i in range(len(A)):
    while tail and A[i] < A[tail[-1]]:
        tail.pop()


    if tail[0] <= i - k:
        tail.popleft()

    if i >= k - 1:
        result.append(A[tail[0]])
        head = deque(tail)
        print(f"Окно {A[i-k+1:i+1]}, tail: {[A[idx] for idx in tail]}, head: {[A[idx] for idx in head]}, минимум: {A[tail[0]]}")

print("\nМинимумы по окнам:", result)
```
