```python
def next_smaller_right(arr):
    n = len(arr)
    result = [-1] * n
    stack = []

    for i in range(n - 1, -1, -1):
        while stack and arr[stack[-1]] >= arr[i]:
            stack.pop()

        if stack:
            result[i] = stack[-1]

        stack.append(i)

    return result


arr = [4, 5, 2, 10, 8, 79, 78, 3, 62, 60, 14, 6, 10]
print(next_smaller_right_index(arr))
```
